```python
import pandas as pd

# 读取 Azimuth 标注文件
az = pd.read_csv("/data5/zhangye/scMRDR/input/PBMC/raw_input/pbmc_azimuth_annotations.csv")

# 提取 l2 cell types
cell_types_l2 = sorted(az["predicted.celltype.l2"].dropna().astype(str).unique())

# 打印结果
print("L2 cell types:")
print(cell_types_l2)
print("n_unique =", len(cell_types_l2))

# 如果还想顺便看每类数量
print("\nL2 cell type counts:")
display(
    az["predicted.celltype.l2"]
    .astype(str)
    .value_counts()
    .rename_axis("cell_type")
    .reset_index(name="n_cells")
)
```
